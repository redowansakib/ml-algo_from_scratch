# Linear Regression

Linear regression is a age old statistical process where we try to fit a line onto some data that describes the trend of the data best. The line itself is called the "Best FIt Line". Now there are differnent ways to fit a line for linear regression but we will be discussing the Mean Square Error or MSE. Where we try to minimize the error.

### Motivation

Linear regression is a technique used to model the relationships between observed variables that might have a linear relationship. The idea behind simple linear regression is to infer the impact of one variable onto some other variable so that we can predict the future occurrence of any variabale by observing relevant other variables.   other.

### Simple Linear Regression
Let us start simple where we will try to execute linear regression for one random input variable $X$ and for one random output variable $Y$. This is called simple linear regression because we want to predict $Y$ from only one variable or we have only one predictor which is $X$. The relationship between $X$ and $Y$ may be representd as,
$$Y = b_0 + b_1X\, + \mathcal{E}$$
Here $\mathcal{E}$ is the error term which is normally distributed. Our aim will be to estimate $b_0$ which is the intercept and $b_1$ the coefficient of feature $X$ or the slope where we minimize $\mathcal{E}$.

For this we assume a line, 
$$\hat{y} = b_0 + b_1x\,$$
Our task in will be to estimate $b_0, b_1$ where  $\,b_0\,$ is the the intercept and $b_1$ is coefficient of input variable. This line predicts $\hat{y}_i$ for a given $x_i$. the prediction ocntains sum amount of error. So, $\hat{y}_i = y_i + \epsilon    $ where $y_i$ is the actual value ans $ \epsilon$ is the error. 

### Cost Function
In order to get the best fit line we should minimize the total error. But only summing up all the error will not provide us with a beeter solution because the positive and negative error will cancel out each other misleading the total scale of error. So intuitively we could use Mean Absoulute Error or **MAE** $(\sum\frac{|\epsilon|}{n}$) , **MSE** ($\frac{\sum\epsilon^2}{n}$), **RSS** $(\sum\epsilon^2)$. MSE is used when the scale is a significant factor like gradient descent and coordiante descent and when scale is not significant RSS gives more easier implementation and mostly used for closed form solution like Oridnary Least Sqaure.

### Ordinary Least Square

Ordinary Least Square or OLS gives a closed form solution where Our loss function is $RSS$ $(\sum\epsilon^2)$ or Residual Sum Squared where we sum up all the errors squared.


$$\mathcal{L} = RSS = \sum_{i=1}^{n}\epsilon^2 = \sum_{i=1}^{n} {(y_i - \hat{y}_i)}^2=  \sum_{i=1}^{n} {(y_i - b_0 + b_1x_i )}^2$$
We have to tweak $b_0$ and $b_1$ in order to get the minimum value for $\mathcal{L}\,$. So, our varible is $b_0$ and $b_1$. In order to minimize our loss function We will do partial deifferntiation of our loss funtion $\mathcal{L}\,$ with respect to $b_0$ and $b_1$ and set the differentiated equation equal to $0$. 
$$ \frac{d \mathcal{L}}{d b_0} = \sum_{i=1}^{n} ( y_i -b_0 + b_1x_i) = 0 \tag{i} $$                                            
$$ \frac{d \mathcal{L}}{d b_1} =\sum_{i=1}^{n}(y_i-b_0 + b_1x_i )x_i = 0 \tag{ii}$$
<br>
To give you an idea about what kind of surface we are dealing with check out the graph below.
<img src="images/parabola.png">

In OLS method we find a closed form solution for each of the coefficnets from partial differential equations of each coefficients. In our simple case we solve for $b_0$ and $b_1$ from equation (i) and (ii) and we get,
$$ b_0 = \bar{y} - b_1 \bar{x} $$
$$ b_1 = \frac{\sum(x - \bar{x})(y - \bar{y})}{\sum{(x - \bar{x})}^2}$$
$$ or , b1 = \frac{n\sum xy - \sum x\sum y}{n\sum x^2 - {(\sum x)}^2} $$
Where $\bar{x}$ and $\bar{y}$ are expected value of $X$ and $Y$. If you are unsure how to reach this solution you can see <a href="https://youtu.be/ewnc1cXJmGA">this video</a>.
<br>

```python
from modules.mylinalg import flatten

def my_simple_linear_regression(x: [list], y: [list]) -> tuple[float, float]:

    x = flatten(x)
    y = flatten(y)
    
    sum_xy = sum([x[i]*y[i] for i in range(len(x))])
    sum_x = sum(x)
    sum_y = sum(y)
    sum_x2 = sum([x[i]**2 for i in range(len(x))])

    b1 = (len(x)*sum_xy - (sum_x * sum_y)) / (len(y)*sum_x2 - sum_x**2)
    b0 = mean(y) - b1*mean(x)

    return b1, b0
```

As a simple trick, for $b_1$ if we divide both the numerator and dinominator with number of observations we get
$$ b_1 = \frac{ \frac{\sum(x - \bar{x})(y - \bar{y})}{n} }{ \frac{\sum{(x - \bar{x})}^2}{n}} $$
The numerator here becomes the covariance between $ x,y$ and the denominator becomes variance of $x$. Ultimately $b_1$ stands for,
$$ b_1 = \frac{cov(x,y)}{var(x)} $$


Let's implemant this simple linear regression in python.

```python
from modules.mystats import cov, mean

def my_simple_linear_regression(x: [list], y: [list]) -> tuple[float, float]:
    
    cov_mat = cov(x, y)
    b1 = cov_mat[0][1] / cov_mat[0][0]
    b0 =  mean(y) - b1*mean(x)

    return b1, b0
```

N.B. <a href="https://github.com/redowansakib/ml-algo_from_scratch/blob/771fd0eddb6491c32963ca918c62defaba4939a6/modules/mystats.py#L109-L125">`cov()`</a> and <a href="https://github.com/redowansakib/ml-algo_from_scratch/blob/771fd0eddb6491c32963ca918c62defaba4939a6/modules/mystats.py#L4-L37">`mean()`</a> are funtions of <a href="https://github.com/redowansakib/ml-algo_from_scratch/blob/771fd0eddb6491c32963ca918c62defaba4939a6/modules/mystats.py">`mystats`</a> which is coded from scratch without help of any advance library. 

Let's compare our `my_simple_linear_regression` with `sklearn.linear_model.LinearRegression` from scikit learn

```python
from sklearn.datasets import make_regression
X, y = make_regression(n_samples=10, n_features=1, noise=8, tail_strength=0.5, random_state=10)
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import StandardScaler
import numpy as np

X_scl = StandardScaler().fit_transform(X.reshape(-1,1))
y_scl = StandardScaler().fit_transform(y.reshape(-1,1))

sk_regression = LinearRegression().fit(X_scl, y_scl)
skb_1, skb_0 = sk_regression.coef_, sk_regression.intercept_

myb_1, myb_0 = my_simple_linear_regression(X_scl.tolist(), y_scl.tolist())
my2b_1, my2b_0 = my_simple_linear_regression(X_scl.flatten().tolist(), y_scl.flatten().tolist())
print(myb_1,myb_0)
print(my2b_1,my2b_0)
```

```python
import matplotlib.pyplot as plt

x_ln = np.linspace(min(X_scl)-(np.mean(X_scl)-min(X_scl))*(10/100), max(X_scl)+(np.mean(X_scl)+max(X_scl)*(10/100)),50)
my_y_ln = myb_0 + myb_1*x_ln
sk_y_ln = skb_0 + skb_1*x_ln

plt.subplots(1,2,figsize =(12,5))

plt.subplot(121)
plt.scatter(X_scl,y_scl)
plt.title(label="my_simple_linear_regression")
plt.xlim(min(X_scl)-(np.mean(X_scl)-min(X_scl))*(10/100),max(X_scl)+(np.mean(X_scl)+max(X_scl)*(10/100)))
plt.ylim(min(y_scl)-(np.mean(y_scl)-min(y_scl))*(10/100),max(y_scl)+(np.mean(y_scl)+max(y_scl)*(10/100)))
plt.plot(x_ln, my_y_ln, color="orange")

plt.subplot(122)
plt.scatter(X_scl,y_scl)
plt.title(label="sk_simple_linear_regression")
plt.xlim(min(X_scl)-(np.mean(X_scl)-min(X_scl))*(10/100),max(X_scl)+(np.mean(X_scl)+max(X_scl)*(10/100)))
plt.ylim(min(y_scl)-(np.mean(y_scl)-min(y_scl))*(10/100),max(y_scl)+(np.mean(y_scl)+max(y_scl)*(10/100)))
plt.plot(x_ln,sk_y_ln, color="red")

plt.show()
```

It seems like our simple linear regression can give a tough competition to Scikit Learn linear regression for one input variable and one output variable. <strong> Success! </strong>

### Multivarirate Linear Regression with OLS

While tackling multivariate linear regression we have to deal with multiple predictors and multiple coefficients which yields multiple and complex equation for each coefficients $b_i$. Lets say we have $n$ predictors (columns) and $m$ observations (rows). In order to estimate all the coefficients we end up with a bunch of $y_i = b_0 + b_1x_{i1} + \cdots +b_nx_{in}\,$ equation. For ease we can represent them as matrix. Let $y$ be the vector matrix for all $y_i\,$ values and $X$ be the matrix for all the $x$ values along with one column of 1's for the bias term $b_0$ and $B$ be the vector matrix of all the coefficents $b_i$. Then our linear model stands for, 

$$y_{m \times 1} = X_{m \times (n+1)} b_{(n+1) \times 1}\, + \mathcal{E}_{m \times 1}$$
<br>
$$\begin{bmatrix}\hat{y}_1\\ \hat{y}_2\\ \vdots \\ \hat{y}_m\end{bmatrix} = 
\begin{bmatrix}1 & x_{11} & x_{12} & \ldots & x_{1n} \\ 
               1 & x_{21} & x_{22} & \ldots & x_{2n}\\ 
               \vdots & \vdots & \vdots & \ddots & \vdots \\ 
               1 & x_{m1} & x_{m2} & \ldots & x_{mn} \end{bmatrix}
               \begin{bmatrix}b_0\\ b_1\\ \vdots \\ b_n\end{bmatrix} + 
               \begin{bmatrix}\epsilon _1\\ \epsilon _2\\ \vdots \\ \epsilon _m\end{bmatrix}$$



Our estimated linear model will be, 

$$\hat{y}_{m \times 1} = X_{m \times (n+1)} b_{(n+1) \times 1}\,$$

Or simply,

$$\hat{y} = X b\,$$

Where $\hat{y}$ are the pedicted values and $b$ will be our estimated values of $B$. Our aim will be as before to minimize the cost function $\sum \epsilon^2_i$ or RSS but this time we will define RSS as $e^Te$ where $e$ stands for error matrix. Our cost function,

$$ \mathcal{L} = \sum_{i=1}^{n} \epsilon^2_i =\epsilon^T\epsilon =  {(y - \hat{y})}^T(y - \hat{y}) = {(y - Xb)}^T(y - Xb) $$

So our aim will be to find $b$ where 
$$ \frac{d\mathcal{L}}{db} = -2X^T(y-Xb) = 0$$

by solving for $b$ from $ \frac{d\mathcal{L}}{db} = 0$ we get,
$$ b = (X^TX)^{-1}X^Ty $$

To see the derivation of the above equation see this <a href="https://www.youtube.com/watch?v=fBaXhzlSO18&t=1s">this video</a>.

Lets implemant out multivariate linear regression.

```python
from modules.mylinalg import *
def multivariate_linear_regression(X, y):
    X = [x.tolist() + [1] for x in X]
    b = matmul(matmul(cholesky_inverse(matmul(transpose(X),X)),transpose(X)),y)
    return [coef[0] for coef in b[:-1]], b[-1]
```

```python
from sklearn.datasets import make_regression
X, y = make_regression(n_samples=50, n_features=5, noise=8, tail_strength=0.5, random_state=10)
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import StandardScaler
import numpy as np

X_scl = StandardScaler().fit_transform(X)
y_scl = StandardScaler().fit_transform(y.reshape(-1,1))

sk_regression = LinearRegression()
sk_regression.fit(X_scl, y_scl)
sk_intercept, sk_coef  = sk_regression.intercept_, sk_regression.coef_

my_coef, my_intercept = multivariate_linear_regression(X_scl, y_scl)

print('SK Intercept ',sk_intercept,'\nOur Intercept',my_intercept)
print('SK Coefficient ', sk_coef,'\nOur Coefficient ',my_coef)
```

```python
from modules.mylinalg import matrix_rank, transpose, cholesky_inverse, matmul, svdvals, to2dim, flatten, dot
from modules.myprocessing import tolist, validate
from modules.mymetrics import r2_score
from modules.mymlalgo import nnls
```

```python
class MyLinearRegression:
    def __init__(self,fit_intercept = True, copy_X=True, positive=False):
        self._set_intercept = fit_intercept
        self._positive = positive
        self._copy_X = copy_X
    
    def fit(self,X,y):

        try:
            self.feature_names_in_ = X.columns
        except AttributeError:
            self.feature_names_in_ = None
   
        X,y = validate(X,y)
        
        if self._copy_X:
            X = X.copy()
        
        
        if len(X) != len(y):
            raise Exception('length of X and y must be same')

        self.rank_ = matrix_rank(X)
        self.singular_ = svdvals(X)
        self.n_features_in_ = len(X[0])
        
        if self._set_intercept:
            X = [[1] + x for x in X]
        
        if self._positive:
            self._b = []
            for i in range(len(y[0])):
                yi = [[t[i]] for t in y]
                b = nnls(X,yi)
                self._b.append(b)
                self._b = transpose(self._b)
        else:
            self._b = matmul(matmul(cholesky_inverse(matmul(transpose(X),X)),transpose(X)),y)
        self.coef_ = transpose(self._b[1:])
        self.intercept_ = self._b[0] 

        return self
    
    def predict(self,X):

        X, = validate(X)
        X = to2dim(X)
        X = [[1] + x for x in X]
        
        y_hat = matmul(X,self._b)

        return y_hat

    def score(self, X, y):
        y_pred = self.predict(X)
        return r2_score(y,y_pred)
        
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import StandardScaler

ss = StandardScaler().set_output(transform='pandas')

X_scl = ss.fit_transform(X)
y = y.reshape(-1,1) if y.ndim ==1 else y

sk_regression = LinearRegression()
sk_regression.fit(X_scl,y)
sk_intercept, sk_coef = sk_regression.intercept_, sk_regression.coef_

my_regression = MyLinearRegression()
my_regression.fit(X_scl, y)
my_intercept, my_coef = my_regression.intercept_, my_regression.coef_

print('SK Intercept ',sk_intercept,'\nOur Intercept',my_intercept)
print('SK Coefficient ', sk_coef,'\nOur Coefficient ',my_coef)
```
