### Ridge Regression

Ridge regression is a mitigation technique for multicollinearity. Multicollinearity in training features causes the coefficients to swell. As a result we end up with large unstable coefficients with large statndard error or confidence interval. These coefficients can be accounted for by making a small change in the loss function. We can add a penalty to the MSE where the penalty. Amongst many penalties Ridge regression uses L2 penalty that takes the norm of all the coefficients. As the loss function is minimized this L2 penalty is also minimized which is the norm of all the coefficients so the coefficients also gets minimized. That is why it is called **L2 Regularaization**. Here, The coefficients shrinks towards 0 but not 0. 

So, our loss function stands,
$$ \displaylines{\mathcal{L}_r = RSS + L2  \\
= \sum_{i=1}^{n} \epsilon^2_i + \sum_{i=0}^mb_i^2 \\
= \sum_{i=1}^{n} {(y_i - \hat{y}_i)}^2 + \sum_{i=0}^mb_i^2 }$$

Because of the addition of the $L2$ (squared pentalty) the smaller wights are penalized less and the larger weights are penalized more.

But here if we penalize the coefficients too much it the coefficients will become very small and we will loose the interpetability of the coefficeints and the predictions will become erroneous as the machines have limited accuracy. To overcome this problem we introdue a scaling factor $\lambda$ that will scale our penalty as we set $\lambda$ accordingly.

$$ \mathcal{L}_r = \sum_{i=1}^{n} {(y_i - \hat{y}_i)}^2 + \lambda\sum_{i=0}^mb_i^2 $$

For any simple ridge regression with only one input variable and one output variable the equation stands for ,
$$ \displaylines{\mathcal{L}_r = RSS + L2 \\
= \sum_{i=1}^{n} \epsilon^2_i + \sum_{i=0}^mb_i^2 \\
= \sum_{i=1}^{n} {(y_i - \hat{y}_i)}^2 + \lambda\sum_{i=0}^mb_i^2 \\
= \sum_{i=1}^{n} {(y_i - b_0 + b_1x_i )}^2 + \lambda(b_0^2+ b_1^2)} $$

And we intend to minimize the $\mathcal{L}_r$ function. From the figure below we can see that how the $L2$ penalty can regularize any coefficient and shrink it towards 0.

<img src="images/ridge.png"> 

From the figure we can see that the minimum point or the depression in $RSS$ was far away from the the (0,0) point wicc represtnts 0 value for both coefficients. The $L2$ penalty has a depression at point (0,0) and as we add $L2$ penalty the deppression of $L2$ moves towards 0 or the $L2$ penalty pulls the minimum point of $RSS$ towards 0. Thus ridge regression regularize any large coefficient by pulling it towards zero.

We can represent the loss function as matrix from,
$$\displaylines{\mathcal{L}_r=\epsilon^T\epsilon + \lambda b^Tb \\ 
=  {(y - \hat{y})}^T(y - \hat{y}) + \lambda b^Tb \\ = 
{(y - Xb)}^T(y - Xb) + \lambda b^Tb}$$


As we differentiate $\mathcal{L}_r$ by $b$ and set it to 0 we get the equation by soloving which we van get $b$,
$$\frac{d\mathcal{L}}{db} = -2X^T(y-Xb) + 2\lambda b = 0$$

By solving the above equation we get,
$$ b = (X^TX+\lambda I)^{-1}X^Ty $$
Where, $I$ is idintity matrix of $n \times n$ size where $n$ = number of columns in $X$

As we can see by subtracting only $\lambda I$ form the $X^TX$ in the original least square solution we can get our regualrized coefficients, we can make our ridge class by only make a small change.

```python
from modules.mylinalg import matsub, identity, scaler_mul, svd_inverse, cholesky_inverse, dot
from modules.myprocessing import validate
```

```python
class MyRidge:
    def __init__(self,alpha=1.0, fit_intercept = True, copy_X=True, positive=False, solver='cholesky'):
        self._alpha = alpha if alpha >= 0 else 0
        self._set_intercept = fit_intercept
        self._positive = positive
        self._copy_X = copy_X
        
        solver_dict = {'cholesky' : cholesky_inverse, 'svd': svd_inverse}
        self._solver = solver_dict.get(solver,None)
        
        if self._solver == None:
            raise Exception('provide cholesky or svd as solver')

        self._gram = None
        self._max_iter = 10000
        self._tol =0.0001
        self._warm_start = False
        self._random_state = 10
        self._b = None
        self._selection = 'cyclic'
    
    def fit(self,X,y):

        try:
            self.feature_names_in_ = X.columns
        except AttributeError:
            self.feature_names_in_ = None
   
        X,y = validate(X,y)
        
        if self._copy_X:
            X = X.copy()

        if len(X) != len(y):
            raise Exception('length of X and y must be same')

        self.rank_ = matrix_rank(X)
        self.singular_ = svdvals(X)
        self.n_features_in_ = len(X[0])
        
        if self._set_intercept:
            X = [[1] + x for x in X]

        if self._positive:
            self._b = [ 0 for i in range(len(X[0]))]

            gram = matmul(transpose(X),X) if not self._gram else self._gram
    
            loop_max_step = 100
            
            while loop_max_step > self._tol:
                max_step = 0
               
                idx_set = set(range(len(self._b)))
                
                j = 0
                iter = 0
                if iter > self._max_iter:
                    break
                
                while idx_set:
                    iter += 1
                    if self._selection == 'random':
                        j = random.choice(idx_set,random_state = self._random_state)
                    idx_set.discard(j)
                    
                    b_j = self._b[j]
                    X_j = [x[j] for x in X]
                    XT_jY = dot(X_j,y)
                    
                    b_k=[]
                    XT_jX_k = []
                    for k in range(len(self._b)):
                        if k != j:
                            b_k.append(self._b[k])
                            XT_jX_k.append(gram[j][k])
    
                    
                    XT_jX_kb_k = dot(XT_jX_k,b_k)
                    rho_j = XT_jY - XT_jX_kb_k
                   
                    z_j = gram[j][j]
    
                    if j == 0 or rho_j > 0:
                        b_j_hat = rho_j/(self._alpha + z_j)
                    else:
                        b_j_hat = 0
    
                    step = abs(b_j_hat - b_j)
                    max_step = max(step, max_step)
                    self._b[j] = b_j_hat
    
                    if self._selection == 'cyclic':
                        j += 1
                loop_max_step = max_step
        else:
            self._b = matmul(matmul(self._solver(matadd(matmul(transpose(X),X),scaler_mul(self._alpha,identity(len(X[0]))))),transpose(X)),y)

        self.coef_ =  transpose(self._b[1:])
        self.intercept_ = [b[0] for b in self._b]

        return self
    
    def predict(self,X):

        X, = validate(X)
        X = to2dim(X)
        X = [[1] + x for x in X]
        
        y_hat = matmul(X,self._b)

        return y_hat

    def score(self, X, y):
        y_pred = self.predict(X)
        return r2_score(y,y_pred)
        
```

```python
from sklearn.datasets import make_regression
X, y = make_regression(n_samples=50, n_features=5, noise=8, tail_strength=0.5, random_state=10)
```

```python
from sklearn.linear_model import Ridge
from modules.mystats import standardize
from modules.mylinalg import to2dim, matrix_rank, svdvals, matmul, matadd, transpose

X_scl = standardize(X)
y = to2dim(y)

alpha = 100

sk_ridge = Ridge(alpha)
sk_ridge.fit(X_scl, y)

my_ridge = MyRidge(alpha=alpha)
my_ridge.fit(X_scl, y)

```

```python
from sklearn.datasets import make_regression
X, y = make_regression(n_samples=50, n_features=5, noise=8, tail_strength=0.5, random_state=10)
X[:,2] = -X[:,2]


from sklearn.linear_model import Ridge
from modules.mystats import standardize
from modules.mylinalg import to2dim, matrix_rank, svdvals, matmul, matadd, transpose

X_scl = standardize(X)
y = to2dim(y)

alpha = 100

sk_ridge = Ridge(alpha,positive=True)
sk_ridge.fit(X_scl, y)

my_ridge = MyRidge(alpha=alpha,positive=True)
my_ridge.fit(X_scl, y)
```
