# Elastic Net

Elastic net is another variant of Linear Regression where both $L1$ and $L2$ penalty is used. The necessity of Elastic Net occured from the drawbacks both ridge and lasso have. 

Ridge lacks sparsity that it can not reduce coefficient to zero maybe close to zero but not to zero. And lasso reduces any coefficient to zero arbitrarily leavoin other collinear coefficient. 

Elastic Net solves both of this problem. It retain the sparsity of Lasso and with that it groups different coefficients by ridge penalty avoiding arbitrary reduction.  

For Elastic Net we define the loss function as,

$$ \displaylines{\mathcal{L} = MSE + L1_{ration} \times L1 + \frac{1}{2}(1-L1_{ration}) \times L2\\
=\frac{1}{2}\sum_{i=1}^{n} \frac{{(y_i - \hat{y}_i)}^2}{n} + \lambda \; L1_{ratio} \sum_{i=0}^m |b_i| + \frac{1}{2}\;\lambda \; (1-L1_{ratio} )\sum_{i=0}^mb_i^2}$$

The extra $\frac{1}{2}$ before $L2$ ratio is for scaling purpose (to discard the $2$ coming from differentiation $L2$).

Differentiating the loss function,

$$ \frac{d\mathcal{L}}{db_j}  = \frac{d}{db_j} \frac{1}{2}\sum_{i=1}^{n} \frac{{(y_i - \hat{y}_i)}^2}{n} + \frac{d}{db_j}\lambda \; L1_{ratio} \sum_{i=0}^m |b_i| + \frac{d}{db_j}\frac{1}{2}\;\lambda \; (1-L1_{ratio} )\sum_{i=0}^mb_i^2$$

For MSE  or $\frac{d}{db_j} \frac{1}{2}\sum_{i=1}^{n} \frac{{(y_i - \hat{y}_i)}^2}{n}$term,
$$\displaylines{\frac{d}{db_j} \frac{1}{2}\sum_{i=1}^{n} \frac{{(y_i - \hat{y}_i)}^2}{n} = \frac{1}{2} \frac{d}{db_j} \sum_{i=1}^{n} \frac{{(y_i - \hat{y}_i)}^2}{n}\\
=- \frac{1}{n}\sum_{i=1}^nx_i^{(j)}(y_i - \sum_{k=0,k \neq j}^m b_kx_i^{(k)}) -\frac{1}{n}\sum_{i=1}^nb_j(x_i^{(j)})^2}
$$

Lets consider  $\sum_{i=1}^nx_i^{(j)}(y_i - \sum_{k=0,k \neq j}^m b_kx_i^{(k)})=\rho_j$ and $2\sum_{i=1}^n b_j (x_i^{(j)})^2 = 2b_j\sum_{i=1}^n(x_i^{(j)})^2 = 2b_j z_j$

So, $$\frac{d }{d b_j}MSE = \frac{-\rho_j + b_j z_j}{n}$$

for $L1$  or $\lambda \; L1_{ratio} \sum_{i=0}^m |b_i|$ term,
$$\frac{dL_1}{dv_j} = \begin{cases}
    -\lambda \; L1_{ratio} & \text{if } b_j < 0 \\
    [-\lambda \; L1_{ratio}, \lambda \;  L1_{ratio}] & \text{if } b_j = 0 \\
    \lambda \; L1_{ratio} & \text{if } b_j > 0
\end{cases}$$

for $L2$  or $\lambda \; (1-L1_{ratio}) \sum_{i=0}^mb_i^2|$ term,

$$ \frac{dL_2}{dv_j} = \lambda \; (1-L1_{ratio})b_j$$

So, differentiation of loss function stands for,
$$ \frac{d\mathcal{L}}{db_j} = \frac{-\rho_j + b_j z_j}{n} + 
\begin{cases}
    -\lambda \; L1_{ratio} & \text{if } b_j < 0 \\
    [-\lambda \; L1_{ratio}, \lambda \;  L1_{ratio}] & \text{if } b_j = 0 \\
    \lambda \; L1_{ratio} & \text{if } b_j > 0
\end{cases} +
\lambda \; (1-L1_{ratio})b_j$$

$$ \frac{d\mathcal{L}}{db_j} = 
\begin{cases}
    \frac{-\rho_j + b_j z_j}{n}-\lambda \; L1_{ratio}+\lambda \; (1-L1_{ratio})b_j & \text{if } b_j < 0 \\
    [\frac{-\rho_j + b_j z_j}{n}-\lambda \; L1_{ratio}+\lambda \; (1-L1_{ratio})b_j, \frac{-\rho_j + b_j z_j}{n}+\lambda \;  L1_{ratio}+\lambda \; (1-L1_{ratio})b_j] & \text{if } b_j = 0 \\
    \frac{-\rho_j + b_j z_j}{n}+\lambda \; L1_{ratio}++\lambda \; (1-L1_{ratio})b_j & \text{if } b_j > 0
\end{cases}$$

For Elastic Net we define the loss function as,

$$ \displaylines{\mathcal{L} = MSE + L1_{ration} \times L1 + \frac{1}{2}(1-L1_{ration}) \times L2\\
=\frac{1}{2}\sum_{i=1}^{n} \frac{{(y_i - \hat{y}_i)}^2}{n} + \lambda \; L1_{ratio} \sum_{i=0}^m |b_i| + \frac{1}{2}\;\lambda \; (1-L1_{ratio} )\sum_{i=0}^mb_i^2}$$

The extra $\frac{1}{2}$ before $L2$ ratio is for scaling purpose (to discard the $2$ coming from differentiation $L2$).

Differentiating the loss function,

$$ \frac{d\mathcal{L}}{db_j}  = \frac{d}{db_j} \frac{1}{2}\sum_{i=1}^{n} \frac{{(y_i - \hat{y}_i)}^2}{n} + \frac{d}{db_j}\lambda \; L1_{ratio} \sum_{i=0}^m |b_i| + \frac{d}{db_j}\frac{1}{2}\;\lambda \; (1-L1_{ratio} )\sum_{i=0}^mb_i^2$$

For MSE  or $\frac{d}{db_j} \frac{1}{2}\sum_{i=1}^{n} \frac{{(y_i - \hat{y}_i)}^2}{n}$term,
$$\displaylines{\frac{d}{db_j} \frac{1}{2}\sum_{i=1}^{n} \frac{{(y_i - \hat{y}_i)}^2}{n} = \frac{1}{2} \frac{d}{db_j} \sum_{i=1}^{n} \frac{{(y_i - \hat{y}_i)}^2}{n}\\
=- \frac{1}{n}\sum_{i=1}^nx_i^{(j)}(y_i - \sum_{k=0,k \neq j}^m b_kx_i^{(k)}) -\frac{1}{n}\sum_{i=1}^nb_j(x_i^{(j)})^2}
$$

Lets consider  $\sum_{i=1}^nx_i^{(j)}(y_i - \sum_{k=0,k \neq j}^m b_kx_i^{(k)})=\rho_j$ and $2\sum_{i=1}^n b_j (x_i^{(j)})^2 = 2b_j\sum_{i=1}^n(x_i^{(j)})^2 = 2b_j z_j$

So, $$\frac{d }{d b_j}MSE = \frac{-\rho_j + b_j z_j}{n}$$

for $L1$  or $\lambda \; L1_{ratio} \sum_{i=0}^m |b_i|$ term,
$$\frac{dL_1}{dv_j} = \begin{cases}
    -\lambda \; L1_{ratio} & \text{if } b_j < 0 \\
    [-\lambda \; L1_{ratio}, \lambda \;  L1_{ratio}] & \text{if } b_j = 0 \\
    \lambda \; L1_{ratio} & \text{if } b_j > 0
\end{cases}$$

for $L2$  or $\lambda \; (1-L1_{ratio}) \sum_{i=0}^mb_i^2|$ term,

$$ \frac{dL_2}{dv_j} = \lambda \; (1-L1_{ratio})b_j$$

So, differentiation of loss function stands for,
$$ \frac{d\mathcal{L}}{db_j} = \frac{-\rho_j + b_j z_j}{n} + 
\begin{cases}
    -\lambda \; L1_{ratio} & \text{if } b_j < 0 \\
    [-\lambda \; L1_{ratio}, \lambda \;  L1_{ratio}] & \text{if } b_j = 0 \\
    \lambda \; L1_{ratio} & \text{if } b_j > 0
\end{cases}+
\lambda \; (1-L1_{ratio})b_j$$

$$ \frac{d\mathcal{L}}{db_j} = 
\begin{cases}
    \frac{-\rho_j + b_j z_j}{n}-\lambda \; L1_{ratio}+\lambda \; (1-L1_{ratio})b_j & \text{if } b_j < 0 \\
    [\frac{-\rho_j + b_j z_j}{n}-\lambda \; L1_{ratio}+\lambda \; (1-L1_{ratio})b_j, \frac{-\rho_j + b_j z_j}{n}+\lambda \;  L1_{ratio}+\lambda \; (1-L1_{ratio})b_j] & \text{if } b_j = 0 \\
    \frac{-\rho_j + b_j z_j}{n}+\lambda \; L1_{ratio}+\lambda \; (1-L1_{ratio})b_j & \text{if } b_j > 0
\end{cases}$$

If we set $\frac{d\mathcal{L}}{db_j}=0$ we get,

$$ b_j = \begin{cases} 
\frac{\rho_j + n \lambda L1_{ratio}}{z_j + n\lambda(1- L1_{ratio})} & \text{if} \; \rho_j <  - n \lambda L1_{ratio}\\
0 & \text{if} \;\rho_j \; \text{in} \; [- n \lambda L1_{ratio}, n \lambda L1_{ratio}]\\
\frac{\rho_j - n \lambda L1_{ratio}}{z_j + n\lambda(1- L1_{ratio})} & \text{if} \; \rho_j >  n \lambda L1_{ratio}
\end{cases} $$

If we want to rearange these equation in matrix form we will get,
$$ \rho_j = X_j^T(Y - X_kb_k) = X_j^TY-X_j^TX_kb_k $$
$$ z_j = X_j^TX_j $$

In order to impleament Elastic Net we just need to change the logic of coordiante descent in `MyLasso` and add one extra input which is $L1_{ratio}$

```python
import random
import numpy as np
from modules.mylinalg import *
from modules.mystats import mysum, matmul,transpose
from modules.myprocessing import *
```

```python

class MyElasticNet:
    def __init__(self,alpha=1.0, 
                 l1_ratio = 0.5,
                 fit_intercept = True, 
                 precompute=False, 
                 copy_X=True, 
                 max_iter=1000, 
                 tol=0.0001, 
                 warm_start=False, 
                 positive=False,
                 random_state=None, 
                 selection='cyclic'):

        
        self._alpha = alpha if alpha >= 0 else 0
        if 0 <= l1_ratio <= 1:
            self._l1_ratio = l1_ratio 
        else:
            raise Exception('l1_ratio must be between 0 and 1')
        self._set_intercept = fit_intercept
        self._gram = precompute
        self._copy_X = copy_X
        self._max_iter = max_iter
        self._tol = tol
        self._warm_start = warm_start
        self._positive = positive
        random.seed(random_state)
        self._b = None
        
        if selection in ('cyclic','random'):
             self._selection = selection
        else:
            raise Exception('selection should be either cyclic or random')
    
    def fit(self,X,y):

        try:
            self.feature_names_in_ = X.columns
        except AttributeError:
            self.feature_names_in_ = None
   
        X,y = validate(X,y)
        
        if self._copy_X:
            X = X.copy()

        if len(X) != len(y):
            raise Exception('length of X and y must be same')

        self.rank_ = matrix_rank(X)
        self.singular_ = svdvals(X)
        self.n_features_in_ = len(X[0])
        
        if self._set_intercept:
            X = [[1] + x for x in X]
        if self._warm_start:
            if not self._b:
                raise Exception('Lasso never fitted')
        else:
            self._b = [ 0 for i in range(len(X[0]))]

        gram = matmul(transpose(X),X) if not self._gram else self._gram

        loop_max_step = 100
        
        while loop_max_step > self._tol:
            max_step = 0
           
            idx_set = set(range(len(self._b)))
            
            j = 0
            iter = 0
            if iter > self._max_iter:
                break
            
            while idx_set:
                iter += 1
                if self._selection == 'random':
                    j = random.choice(idx_set)
                idx_set.discard(j)
                
                b_j = self._b[j]
                X_j = [x[j] for x in X]
                XT_jY = dot(X_j,y)
                
                b_k=[]
                XT_jX_k = []
                for k in range(len(self._b)):
                    if k != j:
                        b_k.append(self._b[k])
                        XT_jX_k.append(gram[j][k])

                
                XT_jX_kb_k = dot(XT_jX_k,b_k)
                rho_j = XT_jY - XT_jX_kb_k
               
                z_j = gram[j][j]

                if j == 0:
                    b_j_hat = rho_j/z_j
                elif rho_j < -self._alpha*len(y)*self._l1_ratio and not self._positive:
                    b_j_hat = (rho_j + self._alpha*len(y)*self._l1_ratio)/(z_j +len(y)*self._alpha * (1-self._l1_ratio))

                elif rho_j > self._alpha*len(y)*self._l1_ratio:
                    b_j_hat = (rho_j - self._alpha*len(y)*self._l1_ratio)/(z_j +len(y)*self._alpha * (1-self._l1_ratio))

                else :
                   b_j_hat = 0

                step = abs(b_j_hat - b_j)
                max_step = max(step, max_step)
                self._b[j] = b_j_hat

                if self._selection == 'cyclic':
                    j += 1
            loop_max_step = max_step
            
        self.coef_ =  self._b[1:]
        self.intercept_ = self._b[0]

        return self
    
    def predict(self,X):

        X, = validate(X)
        X = to2dim(X)
        X = [[1] + x for x in X]
        
        y_hat = matmul(X,self._b)

        return y_hat

    def score(self, X, y):
        y_pred = self.predict(X)
        return r2_score(y,y_pred)
```

```python
from sklearn.datasets import make_regression
X, y = make_regression(n_samples=50, n_features=5, noise=8, tail_strength=0.5, random_state=10)
```

```python
from sklearn.linear_model import ElasticNet
from sklearn.preprocessing import StandardScaler

ss = StandardScaler()
X_scl = ss.fit_transform(X)
y = y.reshape(-1,1) if y.ndim ==1 else y

alpha = 0.1
tolerance = 0.0001
l1_ratio = 0.5

sk_en = ElasticNet(alpha,tol = tolerance, l1_ratio=l1_ratio)
sk_en.fit(X,y)

X = X.tolist()
y = to2dim(y.tolist())

my_en = MyElasticNet(alpha, tol = tolerance,l1_ratio=l1_ratio)
my_en.fit(X,y)
```

```python
print('Sklearn ElasticNet Coefficients',sk_en.coef_)
print('Our ElasticNet Coefficients',my_en.coef_)
```

```python
print('SklearnElasticNet Intercept',sk_en.intercept_)
print('Our ElasticNet Intercept',my_en.intercept_)
```
