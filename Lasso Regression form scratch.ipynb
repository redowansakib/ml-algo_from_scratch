### Lasso Regression

Lassso regression is another regularization method where any coefficients can go to zero. This method is mainly used for **Feature Selectoin** where features with cooefficient zero is removed.
For loss function we use $MSE$ with $L1$ regularization or $\sum{||b||}$ multiplied by a scaler $\lambda$. As a result, the loss function stands for,

$$ \mathcal{L}_l = RSS + L1 =\sum_{i=1}^{n} {(y_i - \hat{y}_i)}^2 + \lambda \sum_{i=0}^m ||b_i|| $$

(*The reson we use MSE is beacuse lasso is implemented by coordinate descent and it is sensitive to scaling. MSE provides a better scaled loss function and MSE multiplied by $\frac{1}{2}$ is another option that simplifies furthur differentiation.*)

Now we have to get to the minimum point of this loss function.

<img src='images/lasso.png'>

Form the image it is evident that $\sum_{i=0}^m ||b_i||$ is not differentiable so we have to implement coordinate descent.

### Coordinate Descent

Coordinate descent is an optimization technique where we have to minimize along a coordinate and then switch to another coordinate to minmize and keep minimizing until we reach the minimum point.

As usually we start by defining our loss function,
$$\mathcal{L}_l = MSE + L1 = \frac{1}{2}\sum_{i=1}^{n} \frac{{(y_i - \hat{y}_i)}^2}{n} + \lambda \sum_{i=0}^m |b_i|$$

And at this point instead of differentiating it, with total $b$ we differentiate only one coordiante $b_j$, $\frac{d \mathcal{L}_l}{d b_j}$

$$\frac{d \mathcal{L}_l}{d b_j} = \frac{d }{d b_j} \frac{1}{2} \frac{(y_i - \hat{y}_i)^2}{n}+ \lambda \sum_{i=0}^m |b_i|)=\frac{1}{2n}\frac{d }{d b_j} (y_i - \hat{y}_i)^2 + \frac{d }{d b_j} \lambda \sum_{i=0}^m |b_i|$$

For $\frac{d }{d b_j} \frac{1}{2n} (y_i - \hat{y}_i)^2$ we get,
$$ \displaylines{ \frac{1}{2n}\frac{d }{d b_j} \sum_{i=1}^{n} {(y_i - \hat{y}_i)}^2 \\
= \frac{1}{2n}\frac{d }{d b_j} \sum_{i=1}^{n} (y_i - \sum_{k=0}^m b_kx_i)^2 \\
=- \frac{1}{n}\sum_{i=1}^nx_i^{(j)} (y_i - \sum_{k=0}^m b_kx_i)^2}$$

Here, $x_i^{(j)}$ is the $j^th$ column associated with $b_j$ coefficient.

$$\displaylines{
= -\frac{1}{n} \sum_{i=1}^nx_i^{(j)}(y_i - \sum_{k=0,k \neq j}^m b_kx_i^{(k)} - b_jx_i^{(j)})\\
= - \frac{1}{n}\sum_{i=1}^nx_i^{(j)}(y_i - \sum_{k=0,k \neq j}^m b_kx_i^{(k)}) -\frac{1}{n}\sum_{i=1}^nb_j(x_i^{(j)})^2}$$

Lets consider  $\sum_{i=1}^nx_i^{(j)}(y_i - \sum_{k=0,k \neq j}^m b_kx_i^{(k)})=\rho_j$ and $2\sum_{i=1}^n b_j (x_i^{(j)})^2 = 2b_j\sum_{i=1}^n(x_i^{(j)})^2 = 2b_j z_j$

So, $$\frac{d }{d b_j}MSE = \frac{-\rho_j + b_j z_j}{n}$$

Now for, $L1$,

$$\frac{d }{d b_j} \lambda \sum_{i=0}^m |b_i| = \lambda \frac{d }{d b_j}  |b_j|$$ 

we can not differentiate $||b_j||$ at 0. For this we will need the concept of subgradient. In subgradient we consider every possible plane as slope bounded at this point. upper bound is the slope defined by previous point and lower bound is defined by the point. 

<center><img src='images\subgradient.png' ></center>


In the image the blue lines are all subgradients where $x = 0 $ for $|x|$ it is bounded by [-1,1]. Any point less than this 0 has a gradient of -1 and and any point greater than 0 has a gradient of 1. 
subgradient of $\lambda |b_j|$ stands for,

$$ \lambda \frac{d }{d b_j}  |b_j| =
\begin{cases}
    -\lambda & \text{if } b_j < 0 \\
    [-\lambda, \lambda] & \text{if } b_j = 0 \\
    \lambda & \text{if } b_j > 0
\end{cases}
$$

so differentitation of cost function is,
$$ \frac{d \mathcal{L}_l}{d b_j} = \frac{\rho_j + b_j z_j}{n} + 
\begin{cases}
    -\lambda & \text{if } b_j < 0 \\
    [-\lambda, \lambda] & \text{if } b_j = 0 \\
    \lambda & \text{if } b_j > 0
\end{cases} $$

$$= \begin{cases}
    \frac{\rho_j + b_j z_j}{n} -\lambda & \text{if } b_j < 0 \\
    \frac{\rho_j + b_j z_j}{n} + [-\lambda, \lambda] & \text{if } b_j = 0 \\
    \frac{\rho_j + b_j z_j}{n} +\lambda & \text{if } b_j > 0
\end{cases}$$

where , $\rho_j = \sum_{i=1}^nx_i^{(j)}(y_i - \sum_{k=0,k \neq j}^m b_kx_i^{(k)})$ and $z_j = \sum_{i=1}^n(x_i^{(j)})^2 $

For minimizing the function $\mathcal{L}$ we set $\frac{d \mathcal{L}_l}{d b_j}=0$

$$ \frac{d \mathcal{L}_l}{d b_j} 
= \begin{cases}
    \frac{\rho_j + b_j z_j}{n} -\lambda & \text{if } b_j < 0 \\
    \frac{\rho_j + b_j z_j}{n} + [-\lambda, \lambda] & \text{if } b_j = 0 \\
    \frac{\rho_j + b_j z_j}{n} +\lambda & \text{if } b_j > 0
\end{cases}\qquad \Huge{= 0} $$

Our new coefficient $\hat{b_j}$ is,

$$ \hat{b_j} = \begin{cases}
(\rho_j + n\lambda_j)/z_j & \text{if} \rho_j < - n\lambda \\
0 & \text{if} \rho_j \;\text{in} \; [n\lambda,- n\lambda] \\
(\rho_j - n\lambda_j)/z_j & \text{if} \rho_j > n\lambda
\end{cases} $$

we are going to update each coefficient with this equation until the maximum change of any coefficient through a loop is less than a threshold vlaue.

If we want to rearange these equation in matrix form we will get,
$$ \rho_j = X_j^T(Y - X_kb_k) = X_j^TY-X_j^TX_kb_k $$
$$ z_j = X_j^TX_j $$

```python
import random
import numpy as np
from modules.mylinalg import *
from modules.mystats import mysum, matmul,transpose
from modules.myprocessing import *
```

```python

class MyLasso:
    def __init__(self,alpha=1.0, 
                 fit_intercept = True, 
                 precompute=False, 
                 copy_X=True, 
                 max_iter=1000, 
                 tol=0.0001, 
                 warm_start=False, 
                 positive=False,
                 random_state=None, 
                 selection='cyclic'):

        
        self._alpha = alpha if alpha >= 0 else 0
        self._set_intercept = fit_intercept
        self._gram = precompute
        self._copy_X = copy_X
        self._max_iter = max_iter
        self._tol = tol
        self._warm_start = warm_start
        self._positive = positive
        random.seed(random_state)
        self._b = None
        
        if selection in ('cyclic','random'):
             self._selection = selection
        else:
            raise Exception('selection should be either cyclic or random')
    
    def fit(self,X,y):

        try:
            self.feature_names_in_ = X.columns
        except AttributeError:
            self.feature_names_in_ = None
   
        X,y = validate(X,y)
        
        if self._copy_X:
            X = X.copy()

        if len(X) != len(y):
            raise Exception('length of X and y must be same')

        self.rank_ = matrix_rank(X)
        self.singular_ = svdvals(X)
        self.n_features_in_ = len(X[0])
        
        if self._set_intercept:
            X = [[1] + x for x in X]
        if self._warm_start:
            if not self._b:
                raise Exception('Lasso never fitted')
        else:
            self._b = [ 0 for i in range(len(X[0]))]

        gram = matmul(transpose(X),X) if not self._gram else self._gram

        loop_max_step = 100
        
        while loop_max_step > self._tol:
            max_step = 0
           
            idx_set = set(range(len(self._b)))
            
            j = 0
            iter = 0
            if iter > self._max_iter:
                break
            
            while idx_set:
                iter += 1
                if self._selection == 'random':
                    j = random.choice(idx_set)
                idx_set.discard(j)
                
                b_j = self._b[j]
                X_j = [x[j] for x in X]
                XT_jY = dot(X_j,y)
                
                b_k=[]
                XT_jX_k = []
                for k in range(len(self._b)):
                    if k != j:
                        b_k.append(self._b[k])
                        XT_jX_k.append(gram[j][k])

                
                XT_jX_kb_k = dot(XT_jX_k,b_k)
                rho_j = XT_jY - XT_jX_kb_k
               
                z_j = gram[j][j]

                if j == 0:
                    b_j_hat = rho_j/z_j
                elif rho_j < -self._alpha*len(y) and not self._positive:
                    b_j_hat = (rho_j + self._alpha*len(y))/z_j

                elif rho_j > self._alpha*len(y):
                    b_j_hat = (rho_j - self._alpha*len(y))/z_j

                else :
                   b_j_hat = 0

                step = abs(b_j_hat - b_j)
                max_step = max(step, max_step)
                self._b[j] = b_j_hat

                if self._selection == 'cyclic':
                    j += 1
            loop_max_step = max_step
            
        self.coef_ =  self._b[1:]
        self.intercept_ = self._b[0]

        return self
    
    def predict(self,X):

        X, = validate(X)
        X = to2dim(X)
        X = [[1] + x for x in X]
        
        y_hat = matmul(X,self._b)

        return y_hat

    def score(self, X, y):
        y_pred = self.predict(X)
        return r2_score(y,y_pred)
```

```python
from sklearn.datasets import make_regression
X, y = make_regression(n_samples=50, n_features=5, noise=8, tail_strength=0.5, random_state=10)
```

```python
from sklearn.linear_model import Lasso
from sklearn.preprocessing import StandardScaler

ss = StandardScaler()

X_scl = ss.fit_transform(X)
y = y.reshape(-1,1) if y.ndim ==1 else y

alpha = 0.1
tolerance = 0.0001

sk_ls = Lasso(alpha,tol = tolerance)
sk_ls.fit(X_scl,y)

my_ls = MyLasso(alpha, tol = tolerance)
my_ls.fit(X_scl,y)
```
